```python
# Q1: Apply Push and Pop operations to demonstrate a stack (plates example)
class Stack:
    """Stack using a Python list (top of the stack = end of the list)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, value):
        self.items.append(value)
        print(f"Push({value})  ->  stack (bottom to top): {self.items}")

    def pop(self):
        if self.is_empty():
            print("Stack Underflow: cannot pop from an empty stack")
            return None
        value = self.items.pop()
        print(f"Pop() -> {value}  ->  stack (bottom to top): {self.items}")
        return value

    def peek(self):
        return None if self.is_empty() else self.items[-1]

    def display(self):
        print("Stack (bottom to top):", self.items if self.items else "empty")


plates = Stack()
print("Placing plates on the stack:")
plates.push(5)
plates.push(15)
plates.push(25)

print("\nRemoving plates:")
plates.pop()      # the plate placed last (25) comes off first
plates.pop()

print("\nPlacing another plate:")
plates.push(35)

print("\nFinal state:")
plates.display()
print("Last In, First Out: 25 was pushed last and popped first.")
```
