```python
# Q6: Evaluation of a postfix expression using a stack
def evaluate_postfix(expression):
    stack = []
    for token in expression.split():
        if token.lstrip("-").isdigit():
            stack.append(int(token))                  # operand: push
        else:
            right = stack.pop()                       # operator: pop two operands
            left = stack.pop()
            if token == "+":
                result = left + right
            elif token == "-":
                result = left - right
            elif token == "*":
                result = left * right
            elif token == "/":
                result = left / right
            elif token == "^":
                result = left ** right
            else:
                raise ValueError(f"Unknown operator: {token}")
            stack.append(result)                      # push the result back
        print(f"token {token:>2}   ->   stack: {stack}")
    return stack.pop()


for expr, infix in (("5 3 + 8 2 - *", "(5 + 3) * (8 - 2)"),
                    ("2 3 1 * + 9 -", "2 + 3 * 1 - 9")):
    print(f"\nPostfix: {expr}      Infix: {infix}")
    print("Result =", evaluate_postfix(expr))
```

```python
# Q5: Push and Pop with an example, and the time complexity of each
import time
class Stack:
    """Stack using a Python list (top of the stack = end of the list)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, value):
        self.items.append(value)
        print(f"Push({value})  ->  stack (bottom to top): {self.items}")

    def pop(self):
        if self.is_empty():
            print("Stack Underflow: cannot pop from an empty stack")
            return None
        value = self.items.pop()
        print(f"Pop() -> {value}  ->  stack (bottom to top): {self.items}")
        return value

    def peek(self):
        return None if self.is_empty() else self.items[-1]

    def display(self):
        print("Stack (bottom to top):", self.items if self.items else "empty")


print("Example: Push(7), Push(9), Pop(), Push(4)")
s = Stack()
s.push(7)
s.push(9)
s.pop()
s.push(4)

print("\nTime complexity: Push = O(1), Pop = O(1)")
print("Both operations only touch the top element, so the cost does not")
print("grow with the stack size. Timing check (time per operation):\n")


class QuietStack:
    def __init__(self):
        self.items = []
    def push(self, v):
        self.items.append(v)
    def pop(self):
        return self.items.pop()


for n in (10_000, 100_000, 1_000_000):
    q = QuietStack()
    t0 = time.perf_counter()
    for i in range(n):
        q.push(i)
    push_time = (time.perf_counter() - t0) / n
    t0 = time.perf_counter()
    for _ in range(n):
        q.pop()
    pop_time = (time.perf_counter() - t0) / n
    print(f"n = {n:>9,}   push: {push_time*1e9:6.0f} ns/op   pop: {pop_time*1e9:6.0f} ns/op")
```

```python
# Q3: Inorder traversal of a binary tree with 10, 20, 30, 40, 50

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def inorder(node, result):
    """Left -> Root -> Right"""
    if node:
        inorder(node.left, result)
        result.append(node.data)
        inorder(node.right, result)
    return result


root = Node(10)
root.left = Node(20)
root.right = Node(30)
root.left.left = Node(40)
root.left.right = Node(50)

print("Inorder traversal (Left, Root, Right):")
print(inorder(root, []))
```
