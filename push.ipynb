```python
# Q2: Push and Pop of the elements 10, 20, 30 and 40
class push:
    """Stack using a Python list (top of the stack = end of the list)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, value):
        self.items.append(value)
        print(f"Push({value})  ->  stack (bottom to top): {self.items}")

    def pop(self):
        if self.is_empty():
            print("Stack Underflow: cannot pop from an empty stack")
            return None
        value = self.items.pop()
        print(f"Pop() -> {value}  ->  stack (bottom to top): {self.items}")
        return value

    def peek(self):
        return None if self.is_empty() else self.items[-1]

    def display(self):
        print("Stack (bottom to top):", self.items if self.items else "empty")


s = push()
print("PUSH PHASE")
for value in (10, 20, 30, 40):
    s.push(value)

print("\nPOP PHASE")
order = []
while not s.is_empty():
    order.append(s.pop())

print("\nOrder of removal:", order)
print("\nExtra pop on the empty stack:")
s.pop()
```
